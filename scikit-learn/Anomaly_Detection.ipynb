```python
import pandas as pd
df = pd.read_csv("TestAnomalyUnlabled.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.iloc[:,1],df.iloc[:,2],1.5, linewidth=0.3)
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
clf = IsolationForest(contamination = 0.05)
clf.fit(df[['temperature_c','humidity_pct','pressure_hpa']])
predictions = clf.predict(df[['temperature_c','humidity_pct','pressure_hpa']])
```

```python
predictions
```

```python
import numpy as np
index = np.where(predictions<0)
index
```
